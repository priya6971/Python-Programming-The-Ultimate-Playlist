````python
import plotly.express as px
data = px.data.iris()
print(data)
````

Built-in interactivity in the Plotly based plots


*   hover
*   zoom
*   legends



**Scatter Plot**

When to use:
*   Identify the relationship among two numeric/continous variables
*   Pattern detection


````python
fig = px.scatter(data, x = 'sepal_width', y = 'sepal_length', color='species')
fig.show()
````

````python
import matplotlib.pyplot as plt
import seaborn as sns

plt.scatter(data['sepal_width'], data['sepal_length'])
plt.xlabel('sepal width')
plt.ylabel('sepal length')
plt.title('IRIS Scatter Plot (Sepal)')

plt.show()
````

**Line Plot**

When to use:


*   Identify the trends (upward/downward)
*   Continuous data
*   Time-based analysis



````python
fig = px.line(data, y='petal_length', title = 'Petal Length Trend')
fig.show()
````

Bar chart

When to use:


*   category-wise comparison
*   aggregated values



````python
fig = px.bar(data, x = 'species', y = 'sepal_length', title = 'Sepal Length by Species')
fig.show()
````

Histogram

When to use:
* Distribution analysis
* Skewness detection

````python
fig = px.histogram(data, x = 'sepal_length', nbins=20, title='Distribution of Sepal Length')
fig.show()
````

````python
fig = px.box(data, x = 'species', y = 'petal_length', title = 'Petal Length Distribution')
fig.show()
````

````python
fig = px.violin(data, x = 'species', y = 'petal_length', box = True, title = 'Petal Length Distribution')
fig.show()
````

Interactive Sliders

````python
df = px.data.gapminder()
df.head(20)
````

````python
fig = px.scatter(
    df,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    animation_frame="year",
    log_x=True,
    title="Gapminder Animated Scatter Plot"
)

fig.show()
````

```python
x = continent
y = pop
animation_frames = year
```

````python
fig = px.bar(df, x = 'continent', y='pop', animation_frame='year', title = 'Population Growth Over Time')
fig.show()
````
